```python
incident = "ransomware"

if incident == "malware":
    print("Isolate infected system")
elif incident == "ransomware":
    print("Disconnect affected systems from network")
elif incident == "brute force":
    print("Block attacking IP and lock accounts")
elif incident == "data breach":
    print("Restrict access and secure data")
elif incident == "ddos":
    print("Enable traffic filtering and rate limiting")
else:
    print("Investigate the incident")
```
