```python
ioc_feed = [
    "192.168.1.100",
    "malware.com",
    "10.0.0.5"
]

internal_logs = [
    "Login from 192.168.1.100",
    "Normal user activity",
    "Connection to unknown server"
]

print("Threat Intelligence Platform Started")

for ioc in ioc_feed:
    for log in internal_logs:
        if ioc in log:
            print("ALERT: IOC Match Found!")
            print("Indicator:", ioc)
            print("Log:", log)
```
