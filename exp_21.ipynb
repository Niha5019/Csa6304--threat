```python
report = "Improve patch management, employee awareness training, and backup recovery."

areas = {
    "Patch Management": ["patch", "software update"],
    "Security Awareness": ["awareness", "training"],
    "Monitoring": ["monitoring", "log analysis"],
    "Backup Strategy": ["backup", "recovery"],
    "Access Control": ["access control", "permissions"]
}

print("Improvement Areas Identified:")

for area, keywords in areas.items():
    if any(word in report.lower() for word in keywords):
        print("-", area)
```
