```python
baseline = {
    "Firewall": True,
    "Antivirus": True,
    "MFA": True,
    "Encryption": True,
    "Backup": True
}

current = {
    "Firewall": True,
    "Antivirus": True,
    "MFA": False,
    "Encryption": True,
    "Backup": False
}

print("Security Baseline Report")

for control, required in baseline.items():
    if required and current.get(control, False):
        print(control, "- COMPLIANT")
    else:
        print(control, "- NON-COMPLIANT")
```
