```python
task = "network scanning"

tools = {
    "Nmap": ["network scanning", "scan ports"],
    "Wireshark": ["packet analysis", "network traffic"],
    "Metasploit": ["penetration testing", "exploit testing"],
    "Nessus": ["vulnerability scanning", "security weaknesses"],
    "Splunk": ["log monitoring", "security monitoring"],
    "Snort": ["intrusion detection", "network attacks"]
}

for tool, keywords in tools.items():
    if any(word in task for word in keywords):
        print("Recommended Security Tool:", tool)
        break
else:
    print("No suitable tool found")
```
