```python
traffic = [
    ("TCP", 80, "malicious request"),
    ("UDP", 53, "normal data"),
    ("TCP", 80, "normal request")
]

protocol = "TCP"
port = 80
pattern = "malicious"

for p, pt, content in traffic:
    if p == protocol and pt == port and pattern in content:
        print("IDS ALERT: Suspicious Traffic Detected")
    else:
        print("Traffic Normal")
```
