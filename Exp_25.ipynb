```python
import hashlib

records = [
    ("Evidence1", 100),
    ("Evidence2", 200),
    ("Evidence3", 300)
]

previous_hash = "0"
previous_time = 0

for data, timestamp in records:
    if timestamp < previous_time:
        print("Chain of Custody Broken")
        break

    current_hash = hashlib.sha256(
        (data + str(timestamp) + previous_hash).encode()
    ).hexdigest()

    previous_hash = current_hash
    previous_time = timestamp
else:
    print("Chain of Custody Verified")
```
