```python
firewall_log = ["192.168.1.50"]
windows_log = ["Failed login from 192.168.1.50"]
vpn_log = ["VPN login attempt from 192.168.1.50"]

for ip in firewall_log:
    if (any(ip in log for log in windows_log) and
            any(ip in log for log in vpn_log)):
        print("Brute Force Attack Detected!")
        print("Suspicious IP Address:", ip)
        print("SIEM Alert Generated")
```
