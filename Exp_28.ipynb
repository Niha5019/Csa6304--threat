```python
incident = input("Enter incident type: ").lower()

if "malware" in incident:
    strategy = "Isolate infected system"
    elif "ransomware" in incident:
        strategy = "Disconnect affected systems from network"
        elif "brute force" in incident:
            strategy = "Block attacking IP and lock affected accounts"
            elif "data breach" in incident:
                strategy = "Restrict access and secure compromised data"
                elif "ddos" in incident:
                    strategy = "Enable traffic filtering and rate limiting"
                    else:
                        strategy = "Investigate incident and apply suitable controls"

                        print("Recommended Containment Strategy:", strategy)
```
