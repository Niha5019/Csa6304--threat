```python
from collections import Counter

logs = [
    "192.168.1.10",
    "192.168.1.20",
    "192.168.1.10",
    "192.168.1.10",
    "192.168.1.10",
    "192.168.1.30"
]

threshold = 3
count = Counter(logs)

for ip, requests in count.items():
    print(ip, ":", requests, "requests")
    if requests > threshold:
        print("Possible DoS Attack detected from", ip)
```
