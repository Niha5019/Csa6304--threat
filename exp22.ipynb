```python
logs = [
    ("admin", "failed"),
    ("admin", "failed"),
    ("admin", "failed"),
    ("admin", "success"),
    ("john", "failed"),
    ("john", "success")
]

user = "admin"
failed = 0
detected = False

for username, status in logs:
    if username == user:
        if status == "failed":
            failed += 1
        elif status == "success":
            if failed >= 3:
                detected = True
                break
            failed = 0

print("Username:", user)

if detected:
    print("Possible brute-force attack detected!")
else:
    print("No brute-force pattern detected.")
```
