```python
description = "Employees receive security awareness training"

text = description.lower()

if "training" in text or "employee" in text or "awareness" in text:
    domain = "People Controls"
elif "door" in text or "cctv" in text:
    domain = "Physical Controls"
elif "firewall" in text or "encryption" in text:
    domain = "Technological Controls"
else:
    domain = "Organizational Controls"

print("Description:", description)
print("Control Domain:", domain)
```
