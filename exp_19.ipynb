```python
events = [
    {"user": "admin", "status": "failed"},
    {"user": "john", "status": "success"},
    {"user": "admin", "status": "failed"},
    {"user": "john", "status": "failed"},
    {"user": "admin", "status": "success"}
]

print("Failed Login Events:")
for event in events:
    if event["status"] == "failed":
        print(event)

counts = {}
for event in events:
    user = event["user"]
    counts[user] = counts.get(user, 0) + 1

print("\nEvent Count By User:")
for user, count in counts.items():
    print(user, ":", count)
```
