```python
roles = {
    "Admin": ["read", "write", "delete"],
    "Employee": ["read"],
    "Manager": ["read", "write"]
}

users = {
    "John": ("Employee", ["read", "write"]),
    "David": ("Admin", ["read", "write", "delete"]),
    "Sara": ("Manager", ["read", "write"])
}

for user, (role, permissions) in users.items():
    excess = set(permissions) - set(roles[role])
    if excess:
        print(user, "has excess permissions:", excess)
    else:
        print(user, "has correct permissions")
```
