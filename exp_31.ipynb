```python
rules = [
    ("192.168.1.10", "ALLOW"),
    ("192.168.1.20", "DENY")
]

def firewall(ip):
    for address, action in rules:
        if ip == address:
            return action
    return "DENY"

test_ips = ["192.168.1.10", "192.168.1.20", "192.168.1.30"]

for ip in test_ips:
    print(ip, ":", firewall(ip))
```
