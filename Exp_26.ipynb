```python
import hashlib

sample = "malware sample"

file_hash = hashlib.sha256(sample.encode()).hexdigest()
print("File Hash:", file_hash)

suspicious = ["keylogger", "delete_files", "steal_data"]

print("Static Analysis:")
for word in suspicious:
    if word in sample:
        print("Suspicious API Found:", word)

print("Dynamic Analysis:")

behavior = ["File access", "Network connection", "Data theft"]

for action in behavior:
    if action == "Data theft":
        print("Suspicious Behavior Detected:", action)
    else:
        print("Behavior:", action)
```
