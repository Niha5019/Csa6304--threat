```python
logs = [
    "Suspicious IP detected",
    "Malware activity found",
    "Unusual login detected",
    "Normal user activity"
]

print("Hypothesis-Based Hunting:")
for log in logs:
    if "Unusual" in log:
        print(log)

print("\nIntelligence-Based Hunting:")
for log in logs:
    if "Suspicious" in log:
        print(log)

print("\nAnalytics-Based Hunting:")
for log in logs:
    if "Malware" in log:
        print(log)
```
