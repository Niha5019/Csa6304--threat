```python
import time

baseline = 2

login_events = [
    ("John", 1),
    ("David", 2),
    ("Unknown", 5),
    ("Sara", 1)
]

print("Security Monitoring Started")

for user, attempts in login_events:
    print("User:", user, "Login Attempts:", attempts)
    if attempts > baseline:
        print("ALERT: Unusual login activity detected!")

time.sleep(1)
print("Monitoring Completed")
```
