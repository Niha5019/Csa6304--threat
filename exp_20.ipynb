```python
import hashlib

sample = b"CreateProcess VirtualAlloc WriteProcessMemory"
apis = ["CreateProcess", "VirtualAlloc", "WriteProcessMemory", "DeleteFile"]

print("Static Analysis")
print("SHA-256:", hashlib.sha256(sample).hexdigest())

for api in apis:
    if api.encode() in sample:
        print("Suspicious API:", api)

print("\nDynamic Analysis")
behavior = ["File created", "Network connection", "Registry modification"]

for action in behavior:
    if action in ["Network connection", "Registry modification"]:
        print("Suspicious behavior:", action)

print("Analysis completed")
```
