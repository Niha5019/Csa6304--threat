```python
import hashlib

records = [
    {"time": 100, "data": "Evidence A", "prev": "0"},
    {"time": 110, "data": "Evidence B", "prev": ""},
    {"time": 120, "data": "Evidence C", "prev": ""}
]

previous_hash = "0"
previous_time = 0

for record in records:
    record["prev"] = previous_hash
    content = str(record["time"]) + record["data"] + record["prev"]
    current_hash = hashlib.sha256(content.encode()).hexdigest()

    if record["time"] < previous_time:
        print("Invalid timestamp chain")
        break

    print(record["data"], "- Chain Verified")
    previous_time = record["time"]
    previous_hash = current_hash
else:
    print("Evidence chain is valid")
```
