```python
rules = [
    {"protocol": "TCP", "port": 80, "pattern": "SQL_INJECTION"},
    {"protocol": "TCP", "port": 443, "pattern": "MALWARE"}
]

packets = [
    {"protocol": "TCP", "port": 80, "data": "SQL_INJECTION detected"},
    {"protocol": "UDP", "port": 53, "data": "SQL_INJECTION detected"},
    {"protocol": "TCP", "port": 443, "data": "MALWARE detected"}
]

for packet in packets:
    alert = False
    for rule in rules:
        if (packet["protocol"] == rule["protocol"]
                and packet["port"] == rule["port"]
                and rule["pattern"] in packet["data"]):
            alert = True

    if alert:
        print("ALERT:", packet)
    else:
        print("Normal traffic")
```
