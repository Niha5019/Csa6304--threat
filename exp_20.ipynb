```python
description = "The attacker sends fake emails to steal passwords"
text = description.lower()

if "fake email" in text or "steal passwords" in text:
    threat = "Phishing"
elif "ddos" in text or "traffic flooding" in text:
    threat = "DoS/DDoS"
elif "ransomware" in text or "virus" in text:
    threat = "Malware"
elif "intercepts communication" in text:
    threat = "Man-in-the-Middle"
else:
    threat = "Unknown Threat"

print("Incident:", description)
print("Threat Category:", threat)
```
