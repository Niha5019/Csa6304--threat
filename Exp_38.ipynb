```python
likelihood = 5
impact = 5

score = likelihood * impact

print("Risk Score:", score)

if score <= 4:
    print("Risk Level: Low")
elif score <= 9:
    print("Risk Level: Medium")
elif score <= 16:
    print("Risk Level: High")
else:
    print("Risk Level: Critical")
```
