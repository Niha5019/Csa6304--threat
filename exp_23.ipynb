```python
whitelist = {"192.168.1.5"}
signatures = ["malware", "sql injection", "port scan"]

traffic = [
    ("192.168.1.5", "malware detected"),
    ("10.0.0.8", "normal web request"),
    ("10.0.0.9", "sql injection attempt"),
    ("10.0.0.10", "port scan detected")
]

for ip, data in traffic:
    if ip in whitelist:
        print(ip, "- Allowed (Trusted Source)")
    elif any(signature in data.lower() for signature in signatures):
        print(ip, "- BLOCKED (Suspicious Traffic)")
    else:
        print(ip, "- Allowed")
```
